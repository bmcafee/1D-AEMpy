```python
df_dev = df[output_columns][df["depth"]==3].drop(columns=["depth"])

# ensure datetime dtype
df_dev['datetime'] = pd.to_datetime(df_dev['datetime'])

fig, ax = plt.subplots(figsize=(16, 4), dpi=150)  
ax.plot(df_dev['datetime'], df_dev['scaled_chla_final06'], linewidth=0.7, alpha=0.9)

fig.tight_layout()
plt.show()
```

```python
# Box plot for scaled_chla_final06 to identify anomalies
import matplotlib.pyplot as plt
import seaborn as sns

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6), dpi=150)

# Box plot
ax1.boxplot(df_dev['scaled_chla_final06'].dropna(), vert=True)
ax1.set_title('Box Plot: scaled_chla_final06', fontsize=14, fontweight='bold')
ax1.set_ylabel('scaled_chla_final06', fontsize=12)
ax1.grid(True, alpha=0.3)

# Add statistical annotations
stats = df_dev['scaled_chla_final06'].describe()
textstr = f"""Statistics:
Mean: {stats['mean']:.4f}
Median: {stats['50%']:.4f}
Std: {stats['std']:.4f}
Min: {stats['min']:.4f}
Max: {stats['max']:.4f}
Q1: {stats['25%']:.4f}
Q3: {stats['75%']:.4f}"""

ax1.text(0.02, 0.98, textstr, transform=ax1.transAxes, fontsize=10,
         verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

# Histogram for distribution
ax2.hist(df_dev['scaled_chla_final06'].dropna(), bins=50, alpha=0.7, edgecolor='black')
ax2.set_title('Histogram: scaled_chla_final06', fontsize=14, fontweight='bold')
ax2.set_xlabel('scaled_chla_final06', fontsize=12)
ax2.set_ylabel('Frequency', fontsize=12)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Print outlier analysis
Q1 = stats['25%']
Q3 = stats['75%']
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df_dev[(df_dev['scaled_chla_final06'] < lower_bound) | 
                 (df_dev['scaled_chla_final06'] > upper_bound)]['scaled_chla_final06']

print(f"\n=== Outlier Analysis ===")
print(f"IQR: {IQR:.4f}")
print(f"Lower fence: {lower_bound:.4f}")
print(f"Upper fence: {upper_bound:.4f}")
print(f"Number of outliers: {len(outliers)}")
print(f"Percentage of outliers: {len(outliers)/len(df_dev)*100:.2f}%")

if len(outliers) > 0:
    print(f"\nOutlier value range: {outliers.min():.4f} to {outliers.max():.4f}")
    print(f"Most extreme outliers:")
    print(f"  Minimum: {outliers.min():.6f}")
    print(f"  Maximum: {outliers.max():.6f}")
```
